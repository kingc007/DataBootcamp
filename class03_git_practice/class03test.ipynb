```python
import pandas as pd
coffee_url = (
    "https://raw.githubusercontent.com/jfkoehler/"
    "bootcamp_fa_26/main/data/coffee_shop_operations_synthetic.csv"
)
coffee = pd.read_csv(coffee_url)
coffee.head()
```

```python
print("Shape:", coffee.shape)
print("Columns:", coffee.columns.tolist())
coffee.info()
```

```python
coffee.describe()
```

```python
coffee["revenue_usd"].head()
```

```python
coffee[["store", "revenue_usd", "transactions"]].head()
```

```python
downtown = coffee[coffee["store"] == "Downtown"]
downtown.head()
```

```python
busy_downtown = coffee[
    (coffee["store"] == "Downtown")
    & (coffee["transactions"] >= 100)
]
busy_downtown.sort_values("revenue_usd", ascending=False)
```

```python
busy_uptown = coffee[coffee["store"] == "Uptown"]
print(busy_uptown)
revenue_high = coffee[coffee["revenue_usd"] >= 1400]
print(revenue_high)
super_busy_uptown = coffee[(coffee["store"] == "Uptown") & (coffee["revenue_usd"] >= 1400)]
super_busy_uptown.sort_values("transactions", ascending = False)
```

```python
print("Mean revenue:", coffee["revenue_usd"].mean())
print("Median revenue:", coffee["revenue_usd"].median())
print("Minimum revenue:", coffee["revenue_usd"].min())
print("Maximum revenue:", coffee["revenue_usd"].max())
print("Total revenue:", coffee["revenue_usd"].sum())
```

```python
store_summary = (
    coffee.groupby("store")
    .agg(
        average_revenue=("revenue_usd", "mean"),
        total_transactions=("transactions", "sum"),
    )
    .round(2)
)
store_summary
```

```python
store_summary = (
    coffee.groupby("store").agg(
        average_revenue = ("revenue_usd", "mean")
    ).round(2)
)
store_summary
```
